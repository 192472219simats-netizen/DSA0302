```python


import nltk
from nltk import CFG
import spacy


grammar = CFG.fromstring("""
S -> NP VP
NP -> PRON | DET NOUN
VP -> VERB NP
PRON -> 'She'
DET -> 'the'
NOUN -> 'boy' | 'book'
VERB -> 'reads'
""")

sentence = "She reads the book".split()

parser = nltk.ChartParser(grammar)

print("CFG PARSE TREE")
print("================")

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()



nlp = spacy.load("en_core_web_sm")

text = "She reads the book"
doc = nlp(text)

print("\nDEPENDENCY PARSING")
print("==================")

print("WORD\tHEAD\tDEPENDENCY")

for token in doc:
    print(
        token.text,
        "\t",
        token.head.text,
        "\t",
        token.dep_
    )



print("\nCOMPARISON")
print("==========")

print("CFG represents hierarchical phrase structure.")
print("Dependency parsing represents relationships between words.")
print("Dependency parsing is better for directly identifying word relationships.")
```

```python
# QUESTION 2
# Top-Down Parsing vs Earley Parsing

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> NP VP
NP -> DET NOUN
NP -> DET NOUN PP
VP -> VERB NP
VP -> VERB NP PP
PP -> PREP NP

DET -> 'the' | 'a'
NOUN -> 'boy' | 'book' | 'table'
VERB -> 'reads'
PREP -> 'on'
""")

sentence = "The boy reads the book".lower().split()


# --------------------------------
# TOP-DOWN PARSING
# --------------------------------

print("TOP-DOWN PARSING")
print("================")

print("Start from S")
print("S -> NP VP")
print("NP -> DET NOUN")
print("VP -> VERB NP")

print("\nInput:")
print(" ".join(sentence))

print("\nTop-down parser starts from the grammar start symbol.")
print("It may need backtracking when a wrong rule is selected.")


# --------------------------------
# EARLEY PARSING
# --------------------------------

print("\nEARLEY PARSING")
print("==============")

parser = EarleyChartParser(grammar)

trees = list(parser.parse(sentence))

print("Number of parse trees:", len(trees))

for tree in trees:
    print(tree)


# --------------------------------
# INCOMPLETE INPUT
# --------------------------------

incomplete_sentence = "The boy reads".lower().split()

print("\nINCOMPLETE INPUT")
print("==============")

print("Input:", " ".join(incomplete_sentence))

print("Earley parser maintains partial parsing states.")
print("This makes it useful for dynamic and incomplete input.")


# --------------------------------
# COMPARISON
# --------------------------------

print("\nCOMPARISON")
print("==========")

print("Top-Down: simple but may require backtracking.")
print("Earley: handles ambiguity and incomplete input better.")
print("Earley is more suitable for dynamic real-time NLP applications.")
```

```python
# QUESTION 3
# CFG vs PCFG vs Neural Parsing
# Ambiguous Sentence

import nltk
from nltk import CFG
from nltk.parse import ChartParser


sentence = "She saw the man with a telescope"

print("INPUT SENTENCE")
print("===============")
print(sentence)


# =====================================
# 1. CFG
# =====================================

print("\n1. CFG PARSING")
print("=============")

grammar = CFG.fromstring("""
S -> NP VP

NP -> PRON
NP -> DET NOUN
NP -> DET NOUN PP

VP -> VERB NP
VP -> VERB NP PP

PP -> PREP NP

PRON -> 'She'
DET -> 'the' | 'a'
NOUN -> 'man' | 'telescope'
VERB -> 'saw'
PREP -> 'with'
""")

parser = ChartParser(grammar)

tokens = sentence.split()

trees = list(parser.parse(tokens))

print("Number of CFG parse trees:", len(trees))

for i, tree in enumerate(trees[:5], 1):
    print("\nParse Tree", i)
    print(tree)


# =====================================
# 2. PCFG
# =====================================

print("\n2. PCFG")
print("=======")

# Simulated probabilities

interpretations = {
    "She used the telescope to see the man": 0.75,
    "The man had the telescope": 0.25
}

for meaning, probability in interpretations.items():
    print(meaning)
    print("Probability:", probability)


best = max(
    interpretations,
    key=interpretations.get
)

print("\nMost probable interpretation:")
print(best)


# =====================================
# 3. NEURAL PARSING
# =====================================

print("\n3. NEURAL PARSING")
print("=================")

print("Neural parser uses contextual information")
print("and learned language patterns.")

# Simulated neural parser output

neural_predictions = {
    "She used the telescope to see the man": 0.88,
    "The man had the telescope": 0.12
}

for meaning, score in neural_predictions.items():
    print(meaning)
    print("Neural score:", score)


neural_best = max(
    neural_predictions,
    key=neural_predictions.get
)

print("\nNeural parser prediction:")
print(neural_best)


# =====================================
# 4. COMPARISON
# =====================================

print("\nCOMPARISON")
print("==========")

print("CFG:")
print("- Generates possible syntactic structures")
print("- Does not rank them naturally")

print("\nPCFG:")
print("- Assigns probabilities to parse trees")
print("- Selects the most probable structure")

print("\nNeural Parser:")
print("- Learns from large datasets")
print("- Uses contextual information")
print("- Handles complex ambiguity better")

print("\nFINAL CONCLUSION")
print("================")

print("Neural parsing is generally most effective")
print("for real-world applications.")
```

```python
# QUESTION 4
# Feature Structures and Subcategorization Frames

# =====================================
# 1. FEATURE STRUCTURES
# =====================================

print("FEATURE STRUCTURES")
print("==================")

subject = {
    "word": "She",
    "person": 3,
    "number": "singular"
}

verb = {
    "word": "likes",
    "person": 3,
    "number": "singular"
}

print("Subject:", subject)
print("Verb:", verb)


# Check subject-verb agreement

if (subject["person"] == verb["person"] and
    subject["number"] == verb["number"]):

    print("Agreement: CORRECT")

else:

    print("Agreement: INCORRECT")


# =====================================
# Incorrect example
# =====================================

subject2 = {
    "word": "They",
    "person": 3,
    "number": "plural"
}

verb2 = {
    "word": "likes",
    "person": 3,
    "number": "singular"
}

print("\nSecond Example")

if (subject2["person"] == verb2["person"] and
    subject2["number"] == verb2["number"]):

    print("Agreement: CORRECT")

else:

    print("Agreement: INCORRECT")


# =====================================
# 2. SUBCATEGORIZATION FRAMES
# =====================================

print("\nSUBCATEGORIZATION FRAMES")
print("========================")

verb_frames = {

    "eat": {
        "pattern": "Subject + Verb + Object",
        "example": "John eats food"
    },

    "give": {
        "pattern": "Subject + Verb + Object + Recipient",
        "example": "John gives Mary a book"
    },

    "sleep": {
        "pattern": "Subject + Verb",
        "example": "John sleeps"
    },

    "put": {
        "pattern": "Subject + Verb + Object + Location",
        "example": "John puts the book on the table"
    }
}


for verb, information in verb_frames.items():

    print("\nVerb:", verb)
    print("Pattern:", information["pattern"])
    print("Example:", information["example"])


# =====================================
# 3. CHECK VERB STRUCTURE
# =====================================

print("\nVERB ARGUMENT CHECK")
print("===================")

verb = "give"

sentence_arguments = [
    "John",
    "gives",
    "Mary",
    "a book"
]

expected_arguments = 4

if len(sentence_arguments) == expected_arguments:
    print("Correct argument structure")

else:
    print("Incorrect argument structure")


# =====================================
# 4. FINAL COMPARISON
# =====================================

print("\nCOMPARISON")
print("==========")

print("Feature Structures:")
print("- Handle grammatical features")
print("- Check number")
print("- Check person")
print("- Check subject-verb agreement")

print("\nSubcategorization Frames:")
print("- Describe verb argument requirements")
print("- Identify required complements")
print("- Handle verb-object relationships")

print("\nCONCLUSION")
print("===========")

print("Feature structures are better for")
print("enforcing grammatical agreement.")

print("Subcategorization frames are better for")
print("representing verb argument structures.")

print("Both should be combined for a complete grammar system.")
```

```python
# QUESTION 5
# Transition-Based vs Graph-Based Dependency Parsing

# =====================================
# SENTENCE
# =====================================

sentence = [
    "She",
    "reads",
    "the",
    "book"
]

print("INPUT SENTENCE")
print("===============")
print(" ".join(sentence))


# =====================================
# 1. TRANSITION-BASED PARSING
# =====================================

print("\n1. TRANSITION-BASED PARSING")
print("===========================")

stack = []
buffer = sentence.copy()
dependencies = []

print("Initial Stack:", stack)
print("Initial Buffer:", buffer)


# SHIFT operation
stack.append(buffer.pop(0))

print("\nSHIFT")
print("Stack:", stack)
print("Buffer:", buffer)


# SHIFT
stack.append(buffer.pop(0))

print("\nSHIFT")
print("Stack:", stack)
print("Buffer:", buffer)


# RIGHT-ARC
dependencies.append(
    ("reads", "She", "subject")
)

print("\nRIGHT-ARC")
print("Dependency:", dependencies[-1])


# SHIFT
stack.append(buffer.pop(0))

print("\nSHIFT")
print("Stack:", stack)
print("Buffer:", buffer)


# SHIFT
stack.append(buffer.pop(0))

print("\nSHIFT")
print("Stack:", stack)
print("Buffer:", buffer)


# Dependency
dependencies.append(
    ("reads", "book", "object")
)

print("\nDEPENDENCY")
print(dependencies[-1])


print("\nTransition-Based Dependencies:")

for head, dependent, relation in dependencies:
    print(
        head,
        "-->",
        dependent,
        "(" + relation + ")"
    )


# =====================================
# 2. GRAPH-BASED PARSING
# =====================================

print("\n2. GRAPH-BASED PARSING")
print("======================")

# Candidate dependency graph

graph = {

    "reads -> She": 0.95,

    "reads -> book": 0.90,

    "book -> the": 0.98,

    "She -> reads": 0.10
}

print("Candidate dependency edges:")

for edge, score in graph.items():
    print(edge, "Score:", score)


# Select high scoring edges

best_edges = {
    edge: score
    for edge, score in graph.items()
    if score >= 0.80
}

print("\nBest Dependency Graph:")

for edge, score in best_edges.items():
    print(edge, "Score:", score)


# =====================================
# 3. COMPARISON
# =====================================

print("\nCOMPARISON")
print("==========")

print("\nTransition-Based:")
print("- Uses stack and buffer")
print("- Makes local decisions")
print("- Very fast")
print("- Suitable for real-time parsing")

print("\nGraph-Based:")
print("- Creates candidate dependency graph")
print("- Scores possible dependency structures")
print("- Selects the highest scoring structure")
print("- Usually provides stronger global decisions")


# =====================================
# 4. PERFORMANCE
# =====================================

print("\nPERFORMANCE")
print("===========")

sentence_length = len(sentence)

transition_operations = sentence_length * 2

print("Sentence length:", sentence_length)
print("Approximate transition operations:",
      transition_operations)

print("\nTransition-based parsing is generally")
print("faster and more suitable for large-scale")
print("real-time applications.")

print("\nGraph-based parsing considers global")
print("relationships but can require more computation.")


# =====================================
# FINAL RESULT
# =====================================

print("\nFINAL CONCLUSION")
print("================")

print("For large-scale applications where speed")
print("is important, transition-based parsing")
print("is generally more suitable.")

print("Graph-based parsing is useful when global")
print("dependency accuracy is more important.")
```
