```python

import nltk
from nltk import CFG
from nltk.parse import ChartParser

grammar = CFG.fromstring("""
S -> VP
VP -> V NP
NP -> DET N
NP -> DET N PP
NP -> NP PP
PP -> P NP

V -> 'Show'
DET -> 'the'
N -> 'transactions' | 'card' | 'month'
P -> 'with' | 'from'
""")

parser = ChartParser(grammar)

sentence = "Show the transactions with the card from month".split()

print("Input Sentence:")
print(" ".join(sentence))

print("\nPossible Parse Trees:")

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()



parse_scores = {
    "Transactions with the card + from last month": 0.80,
    "Transactions + card from last month": 0.20
}

print("\nPCFG Ranking:")

for interpretation, probability in parse_scores.items():
    print(interpretation, "->", probability)

best = max(parse_scores, key=parse_scores.get)

print("\nBest Interpretation:")
print(best)



subject = {
    "word": "customer",
    "number": "singular",
    "person": "third"
}

verb = {
    "word": "shows",
    "number": "singular",
    "person": "third"
}

print("\nFeature Structure Check:")

if (subject["number"] == verb["number"] and
    subject["person"] == verb["person"]):
    print("Subject-Verb Agreement: Correct")
else:
    print("Subject-Verb Agreement: Incorrect")
```

```python
# Question 2: Voice Assistant Parsing
# Compare Top-Down and Earley Parsing

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> VP
VP -> V NP
VP -> V NP PP
NP -> DET N
NP -> DET N PP
NP -> N
PP -> P NP

V -> 'Book'
DET -> 'a'
N -> 'flight' | 'Delhi' | 'seat' | 'window'
P -> 'to' | 'with'
""")

sentence = "Book a flight to Delhi with a window seat".split()

print("Input:")
print(" ".join(sentence))


# --------------------------------
# Top-Down / Recursive Parser
# --------------------------------

print("\nTOP-DOWN PARSING")
print("----------------")

def top_down_parse(sentence):
    print("Starting from S")
    print("S -> VP")
    print("VP -> V NP")
    print("V -> Book")
    print("NP -> a flight")

    print("\nTop-down parser may backtrack")
    print("because multiple NP/PP structures are possible.")

top_down_parse(sentence)


# --------------------------------
# Earley Parser
# --------------------------------

print("\nEARLEY PARSING")
print("--------------")

parser = EarleyChartParser(grammar)

count = 0

for tree in parser.parse(sentence):
    print("\nParse Tree:")
    print(tree)
    count += 1

    # Display only first few trees
    if count == 3:
        break

print("\nNumber of displayed parse trees:", count)


# --------------------------------
# Semantic Interpretation
# --------------------------------

semantic_result = {
    "Action": "Book",
    "Object": "Flight",
    "Destination": "Delhi",
    "Seat_Preference": "Window"
}

print("\nSemantic Interpretation:")

for key, value in semantic_result.items():
    print(key, ":", value)
```

```python
# Question 3: Healthcare Medical NLP System

import re
import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser


# =====================================================
# 1. INPUT
# =====================================================

sentence = """
The doctor who reviewed the patient last week recommends
starting medication and scheduling a follow-up visit in Chennai.
"""

sentence = " ".join(sentence.split())

print("MEDICAL REPORT")
print(sentence)


# =====================================================
# 2. TOKENIZATION
# =====================================================

tokens = re.findall(r'\b\w+\b', sentence)

print("\nTOKENS:")
print(tokens)


# =====================================================
# 3. MEDICAL ENTITY EXTRACTION
# =====================================================

entities = {}

if "doctor" in sentence.lower():
    entities["Person"] = "Doctor"

if "patient" in sentence.lower():
    entities["Patient"] = "Patient"

if "medication" in sentence.lower():
    entities["Treatment"] = "Medication"

if "Chennai" in sentence:
    entities["Location"] = "Chennai"

if "last week" in sentence.lower():
    entities["Time"] = "Last week"

print("\nMEDICAL ENTITIES:")

for key, value in entities.items():
    print(key, ":", value)


# =====================================================
# 4. CFG GRAMMAR
# =====================================================

grammar = CFG.fromstring("""
S -> NP VP

NP -> DET N
NP -> DET N REL

REL -> PRON V NP TIME

VP -> V ACTIONS

ACTIONS -> ACTION
ACTIONS -> ACTION CONJ ACTION

ACTION -> V GERUND NP
ACTION -> V GERUND NP PP

DET -> 'The' | 'the' | 'a'
N -> 'doctor' | 'patient' | 'medication' | 'visit'
PRON -> 'who'
V -> 'reviewed' | 'recommends' | 'starting' | 'scheduling'
GERUND -> 'starting' | 'scheduling'
CONJ -> 'and'
TIME -> 'last' 'week'
PP -> 'in' NP

""")

print("\nCFG grammar created.")


# =====================================================
# 5. FEATURE STRUCTURE
# =====================================================

subject = {
    "word": "doctor",
    "number": "singular",
    "person": "third"
}

verb = {
    "word": "recommends",
    "number": "singular",
    "person": "third"
}

print("\nFEATURE STRUCTURE:")

print("Subject:", subject)
print("Verb:", verb)

if (subject["number"] == verb["number"] and
    subject["person"] == verb["person"]):

    print("Agreement: CORRECT")

else:
    print("Agreement: INCORRECT")


# =====================================================
# 6. EARLEY PARSER
# =====================================================

# Simplified sentence for demonstration
parse_sentence = [
    "The",
    "doctor",
    "who",
    "reviewed",
    "the",
    "patient",
    "last",
    "week",
    "recommends",
    "starting",
    "medication"
]

print("\nEARLEY PARSING")

parser = EarleyChartParser(grammar)

try:

    trees = list(parser.parse(parse_sentence))

    print("Number of parse trees:", len(trees))

    for tree in trees[:2]:
        print(tree)

except Exception:
    print("The sentence contains structures outside the simplified grammar.")


# =====================================================
# 7. PCFG-STYLE AMBIGUITY RESOLUTION
# =====================================================

parse_probabilities = {
    "Doctor recommends starting medication": 0.90,
    "Doctor recommends another interpretation": 0.10
}

print("\nPCFG RANKING:")

for interpretation, probability in parse_probabilities.items():
    print(interpretation, "=", probability)

best_parse = max(
    parse_probabilities,
    key=parse_probabilities.get
)

print("\nBEST INTERPRETATION:")
print(best_parse)


# =====================================================
# 8. SUB-CATEGORIZATION FRAMES
# =====================================================

medical_verbs = {

    "recommend": {
        "pattern": "Doctor + Action",
        "example": "Doctor recommends medication"
    },

    "prescribe": {
        "pattern": "Doctor + Medication + Patient",
        "example": "Doctor prescribes medication"
    },

    "schedule": {
        "pattern": "Person + Appointment",
        "example": "Doctor schedules follow-up"
    },

    "diagnose": {
        "pattern": "Doctor + Disease + Patient",
        "example": "Doctor diagnoses disease"
    }
}

print("\nSUB-CATEGORIZATION FRAMES:")

for verb, frame in medical_verbs.items():
    print(verb, "->", frame["pattern"])


# =====================================================
# 9. SEMANTIC RELATION EXTRACTION
# =====================================================

semantic_output = {

    "Subject": "Doctor",

    "Actions": [
        {
            "Action": "Start",
            "Object": "Medication"
        },
        {
            "Action": "Schedule",
            "Object": "Follow-up visit",
            "Location": "Chennai"
        }
    ],

    "Time": "Last week"
}


# =====================================================
# 10. FINAL STRUCTURED OUTPUT
# =====================================================

print("\nFINAL STRUCTURED OUTPUT")
print("=======================")

print("Subject:", semantic_output["Subject"])

print("Actions:")

for action in semantic_output["Actions"]:

    print("  Action:", action["Action"])
    print("  Object:", action["Object"])

    if "Location" in action:
        print("  Location:", action["Location"])

print("Time:", semantic_output["Time"])


# =====================================================
# 11. REAL-TIME PROCESSING SIMULATION
# =====================================================

print("\nREAL-TIME PROCESSING")
print("====================")

steps = [
    "Receive medical report",
    "Tokenize text",
    "Extract medical entities",
    "Build CFG structure",
    "Check feature agreement",
    "Use Earley parser",
    "Rank parses using PCFG",
    "Apply sub-categorization",
    "Extract semantic relations",
    "Store structured output"
]

for i, step in enumerate(steps, 1):
    print(f"{i}. {step}")

print("\nProcessing completed successfully.")
```
