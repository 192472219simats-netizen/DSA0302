```python
queries = {
    "Q1": {
        "query": "Activate international roaming for my number.",
        "representation": "ACTIVATE(Roaming, Customer)",
        "actual": "Activate Roaming",
        "predicted": "Activate Roaming"
    },
    "Q2": {
        "query": "Deactivate caller tune service.",
        "representation": "DEACTIVATE(CallerTune, Customer)",
        "actual": "Deactivate Caller Tune",
        "predicted": "Activate Caller Tune"
    },
    "Q3": {
        "query": "Check my data balance.",
        "representation": "QUERY(DataBalance, Customer)",
        "actual": "Query Data Balance",
        "predicted": "Query Data Balance"
    },
    "Q4": {
        "query": "Enable 5G service.",
        "representation": "ACTIVATE(5GService, Customer)",
        "actual": "Activate 5G Service",
        "predicted": "Activate 5G Service"
    }
}

correct = 0

for qid, data in queries.items():
    print("\n", qid)
    print("Query:", data["query"])
    print("Representation:", data["representation"])
    print("Actual:", data["actual"])
    print("Predicted:", data["predicted"])

    if data["actual"] == data["predicted"]:
        print("Result: Correct")
        correct += 1
    else:
        print("Result: ERROR")

accuracy = correct / len(queries) * 100
print("\nAccuracy:", accuracy, "%")
```

```python
machines = {
    "M1": "Active",
    "M2": "Active",
    "M3": "Maintenance",
    "M4": "Active"
}

produces = {
    "M1": ["Gear"],
    "M2": ["Wheel"],
    "M4": ["Bolt"]
}

producing = []
available = []

print("Machine Facts:")

for machine, status in machines.items():
    print(machine, "->", status)

    if status == "Active":
        producing.append(machine)

        if machine in produces:
            for product in produces[machine]:
                available.append(product)

    elif status == "Maintenance":
        print(machine, "-> Not Producing")

print("\nProducing Machines:", producing)
print("Available Products:", available)

if "Gear" in available:
    print("Gear production: Available")
else:
    print("Gear production: Not Available")

if machines["M3"] == "Maintenance":
    print("M3 maintenance affects only products produced by M3.")
```

```python
sentences = [
    {
        "sentence": "Doctor prescribed medicine to patient.",
        "roles": {
            "Doctor": "Agent",
            "Medicine": "Theme",
            "Patient": "Recipient"
        }
    },
    {
        "sentence": "Patient reported severe headache.",
        "roles": {
            "Patient": "Experiencer",
            "Headache": "Symptom"
        }
    },
    {
        "sentence": "Nurse monitored patient continuously.",
        "roles": {
            "Nurse": "Agent",
            "Patient": "Theme"
        }
    },
    {
        "sentence": "Medicine reduced blood pressure.",
        "roles": {
            "Medicine": "Cause",
            "Blood Pressure": "Theme"
        }
    }
]

for item in sentences:
    print("\nSentence:", item["sentence"])
    print("Semantic Roles:")

    for entity, role in item["roles"].items():
        print(entity, "->", role)
```
