```python
# Q1: Constraint-Based Coreference Resolution

text = """John and Mary went to the park.
He brought a ball.
She wanted to play with it.
The dog chased him excitedly.
Finally, they all went home."""

# Entities
entities = ["John", "Mary", "park", "ball", "dog"]

# Possible antecedents
references = {
    "He": ["John", "Mary"],
    "She": ["John", "Mary"],
    "it": ["ball", "park"],
    "him": ["John", "Mary", "dog"],
    "they": [
        "John + Mary",
        "John + Mary + dog"
    ],
    "all": [
        "John + Mary",
        "John + Mary + dog"
    ]
}

print("POSSIBLE ANTECEDENTS")
for pronoun, candidates in references.items():
    print(pronoun, "->", candidates)

# Constraint-based resolution
resolved = {
    "He": "John",
    "She": "Mary",
    "it": "ball",
    "him": "John",
    "they": "John + Mary + dog",
    "all": "John + Mary + dog"
}

print("\nFINAL COREference RESOLUTION")
for pronoun, antecedent in resolved.items():
    print(pronoun, "->", antecedent)

# Constraints
print("\nCONSTRAINTS")

print("1. Gender and Number Agreement:")
print("He -> John")
print("She -> Mary")
print("him -> John")
print("they -> John + Mary + dog")

print("\n2. Recency:")
print("it -> ball because ball is recently mentioned.")

print("\n3. Semantic Compatibility:")
print("She wanted to play with it -> it = ball")
print("Dog chased him -> him = John")

print("\n4. Coherence:")
print("John -> He -> him")
print("Mary -> She")
print("Ball -> it")
print("John + Mary + Dog -> they/all")

# Final resolved paragraph
resolved_text = (
    "John and Mary went to the park. "
    "John brought a ball. "
    "Mary wanted to play with the ball. "
    "The dog chased John excitedly. "
    "Finally, John, Mary, and the dog all went home."
)

print("\nRESOLVED PARAGRAPH:")
print(resolved_text)
```

```python
# Q2: Dialogue Generation
# Required Dialog Act: Advise + Encourage

user_input = "I have an important exam tomorrow but I’m not able to concentrate."

# Discourse planning
print("DISCOURSE PLANNING")
print("1. Identify problem: Cannot concentrate")
print("2. Important entity: exam")
print("3. Dialogue acts: Advice + Encourage")
print("4. Use Cause-Effect relation")
print("5. Include keywords: focus, break, confident")
print("6. Maintain positive tone")
print("7. Generate 2-3 sentences")

# Three possible responses
responses = [

    "Since your exam is tomorrow, take a short break first and then return "
    "to your studies with a clear focus. You can concentrate better in small "
    "study sessions, so stay confident and do your best.",

    "Because your exam is tomorrow, take a short break and remove distractions "
    "so you can focus better. You have prepared for this, so stay confident and keep going.",

    "Your exam is important, so take a short break and then study in small "
    "focused sessions to improve your concentration. This can help you focus "
    "better, and you should stay confident about your preparation."
]

print("\nTHREE POSSIBLE RESPONSES")

for i, response in enumerate(responses, 1):
    print("\nResponse", i)
    print(response)

# Evaluation
print("\nEVALUATION")

for i, response in enumerate(responses, 1):

    sentences = response.count(".")
    has_focus = "focus" in response.lower()
    has_break = "break" in response.lower()
    has_confident = "confident" in response.lower()

    print("\nResponse", i)
    print("Sentences:", sentences)
    print("Contains focus:", has_focus)
    print("Contains break:", has_break)
    print("Contains confident:", has_confident)

print("\nBEST RESPONSE")
print(responses[1])

print("\nREASON:")
print("It gives practical advice.")
print("It motivates the student.")
print("It contains focus, break and confident.")
print("It maintains exam and concentration context.")
print("It has a clear cause-effect relation.")
```

```python
# Q3: Constraint-Based Word Sense Disambiguation

sentence = (
    "The bank by the river flooded after the storm, "
    "but it was saved by quick action."
)

print("SOURCE SENTENCE:")
print(sentence)

# ------------------------------------------------
# Step 1: Identify ambiguous word
# ------------------------------------------------

word = "bank"

possible_meanings = [
    "Financial Institution",
    "Riverbank"
]

print("\nAMBIGUOUS WORD:")
print(word)

print("\nPOSSIBLE MEANINGS:")
for meaning in possible_meanings:
    print("-", meaning)

# ------------------------------------------------
# Step 2: Apply contextual constraints
# ------------------------------------------------

print("\nCONSTRAINT ANALYSIS")

print("Context phrase: 'by the river'")
print("Constraint 1: Location")
print("The word bank is associated with a river.")

print("\nConstraint 2: Semantic Compatibility")
print("A riverbank can flood.")
print("A financial institution is less suitable in this context.")

print("\nConstraint 3: Coherence")
print("The meaning must remain consistent with 'flooded'.")

# Final word sense
resolved_meaning = "Riverbank"

print("\nRESOLVED WORD SENSE:")
print(word, "->", resolved_meaning)

# ------------------------------------------------
# Step 3: Predicate Logic
# ------------------------------------------------

print("\nPREDICATE LOGIC REPRESENTATION")

logic = [
    "bank(b)",
    "river(r)",
    "location(b, r)",
    "flood(b)",
    "after(flood(b), storm)",
    "quick_action(a)",
    "saved(b, a)"
]

for predicate in logic:
    print(predicate)

print("\nComplete Representation:")
print(
    "bank(b) AND river(r) AND location(b,r) AND "
    "flood(b) AND after(flood(b),storm) AND "
    "quick_action(a) AND saved(b,a)"
)

# ------------------------------------------------
# Step 4: Discourse Relation
# ------------------------------------------------

print("\nDISCOURSE RELATION")
print("Clause 1: The riverbank flooded after the storm.")
print("Clause 2: It was saved by quick action.")
print("Relation: CONTRAST")

# ------------------------------------------------
# Step 5: RST-style discourse tree
# ------------------------------------------------

print("\nRST-STYLE DISCOURSE TREE")

print("                 CONTRAST")
print("                /        \\")
print("               /          \\")
print("       CLAUSE 1           CLAUSE 2")
print("           |                   |")
print("   Riverbank flooded     Bank was saved")
print("     after storm          by quick action")

# ------------------------------------------------
# Step 6: Generate target sentence
# ------------------------------------------------

target_sentence = (
    "The riverbank flooded after the storm, "
    "but quick action saved it from further damage."
)

print("\nGENERATED SENTENCE:")
print(target_sentence)

# ------------------------------------------------
# Step 7: Advantages
# ------------------------------------------------

print("\nADVANTAGES OF CONSTRAINT-BASED APPROACH")

advantages = [
    "Reduces word ambiguity",
    "Uses contextual information",
    "Uses semantic compatibility",
    "Maintains discourse coherence",
    "Preserves important entities",
    "Improves NLP accuracy"
]

for i, advantage in enumerate(advantages, 1):
    print(i, ".", advantage)
```
