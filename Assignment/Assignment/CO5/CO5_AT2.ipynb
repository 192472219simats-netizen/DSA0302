```python
text = "Ravi met Arun at the library. He borrowed a book and later returned it."

# Entities identified from the first sentence
entities = ["Ravi", "Arun"]

# Reference resolution
# "He" refers to Ravi based on discourse context
he_antecedent = "Ravi"

# "it" refers to the previously mentioned object "book"
it_antecedent = "the book"

resolved_text = (
    "Ravi met Arun at the library. "
    + he_antecedent +
    " borrowed a book and later returned "
    + it_antecedent + "."
)

print("Original:")
print(text)

print("\nResolved Discourse:")
print(resolved_text)
```

```python
sentences = [
    "The roads were flooded after heavy rainfall.",
    "Therefore, schools were closed for the day.",
    "Students attended classes online."
]

relations = {
    "Sentence 1 -> Sentence 2": "Cause-Effect",
    "Sentence 2 -> Sentence 3": "Result / Sequence"
}

print("Discourse Structure:\n")

print("S1:", sentences[0])
print("   ↓ Cause")
print("S2:", sentences[1])
print("   ↓ Result / Sequence")
print("S3:", sentences[2])

print("\nRelations:")
for relation, value in relations.items():
    print(relation, "=", value)
```

```python
conversation = [
    ("User", "Can you book a train ticket for me?"),
    ("Agent", "Sure, where would you like to travel?"),
    ("User", "I want to go to Chennai."),
    ("Agent", "Your ticket has been booked.")
]

def dialogue_act(text):
    text_lower = text.lower()

    if "can you" in text_lower or "book" in text_lower:
        return "Request"
    elif "where" in text_lower or "?" in text_lower:
        return "Question"
    elif "i want" in text_lower or "chennai" in text_lower:
        return "Inform"
    elif "booked" in text_lower:
        return "Confirmation"
    else:
        return "Action"

for speaker, utterance in conversation:
    act = dialogue_act(utterance)
    print(speaker + ":", utterance)
    print("Dialogue Act:", act)
    print()
```

```python
semantic_input = {
    "Action": "buy",
    "Agent": "student",
    "Object": "book",
    "Tense": "past"
}

# Lexical selection
verb = semantic_input["Action"]
subject = semantic_input["Agent"]
obj = semantic_input["Object"]

# Surface realization for past tense
if semantic_input["Tense"] == "past":
    verb = "bought"

sentence = subject.capitalize() + " " + verb + " a " + obj + "."

print("Semantic Representation:")
print(semantic_input)

print("\nGenerated Sentence:")
print(sentence)
```

```python
source = "The boy is playing football."

# Step 1: Source analysis
interlingua = {
    "ACTION": "PLAY",
    "AGENT": "BOY",
    "OBJECT": "FOOTBALL",
    "TENSE": "PRESENT",
    "ASPECT": "CONTINUOUS"
}

# Step 2: Candidate translations
candidates = {
    "மாணவன் கால்பந்து விளையாடுகிறான்.": 0.92,
    "சிறுவன் கால்பந்து விளையாடுகிறான்.": 0.88,
    "மாணவன் கால்பந்து விளையாடினான்.": 0.55
}

# Step 3: Statistical scoring
best_translation = max(candidates, key=candidates.get)
best_score = candidates[best_translation]

print("Source Sentence:")
print(source)

print("\nInterlingua Representation:")
print(interlingua)

print("\nCandidate Translations:")
for sentence, score in candidates.items():
    print(sentence, "-> Score:", score)

print("\nFinal Translation:")
print(best_translation)

print("Best Score:", best_score)
```
