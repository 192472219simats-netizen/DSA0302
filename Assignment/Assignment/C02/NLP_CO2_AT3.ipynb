```python
#QUESTION 1

import re
from nltk.stem import PorterStemmer

# Create Porter Stemmer
stemmer = PorterStemmer()

# Biomedical corpus
text = """
Infection is a common clinical condition.
The infectious disease was treated successfully.
The patient was infected with a bacterial infection.
Researchers study infection and infectious diseases.
The patient recovered from the infection.
Doctors identified an infectious agent.
The infected patient received treatment.
The study investigates new methods to prevent infection.
"""

# -------------------------------
# 1. PREPROCESSING
# -------------------------------

# Convert text to lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r'[^a-z\s]', '', text)

# Convert text into words
words = text.split()

print("PREPROCESSED WORDS")
print("------------------")
print(words)


# -------------------------------
# 2. APPLY PORTER STEMMER
# -------------------------------

print("\nWORD AND PORTER STEM")
print("--------------------")

for word in words:
    stem = stemmer.stem(word)
    print(f"{word:15} -> {stem}")


# -------------------------------
# 3. ANALYZE IMPORTANT WORDS
# -------------------------------

test_words = [
    "infect",
    "infected",
    "infection",
    "infectious"
]

print("\nMORPHOLOGICAL ERROR ANALYSIS")
print("----------------------------")

for word in test_words:

    stem = stemmer.stem(word)

    if word == "infect":
        observation = "Root word"

    elif word == "infected":
        observation = "Inflectional suffix '-ed' removed"

    elif word == "infection":
        observation = "Derivational suffix '-ion' removed"

    elif word == "infectious":
        observation = "Derivational information lost"

    print(
        f"{word:12} -> {stem:10} : {observation}"
    )


# -------------------------------
# 4. IDENTIFY STEMMING ERRORS
# -------------------------------

print("\nSTEMMING ERRORS")
print("--------------")

for word in test_words:

    stem = stemmer.stem(word)

    if word == "infectious" and stem != "infect":
        print(
            f"{word} -> {stem} : Incorrect/unnatural stem"
        )

    elif word == "infection":
        print(
            f"{word} -> {stem} : Derivational information removed"
        )

    elif word == "infected":
        print(
            f"{word} -> {stem} : Inflectional information removed"
        )

    else:
        print(
            f"{word} -> {stem} : No major stemming error"
        )


# -------------------------------
# 5. INFLECTIONAL VS DERIVATIONAL
# -------------------------------

print("\nINFLECTIONAL VS DERIVATIONAL MORPHOLOGY")
print("----------------------------------------")

print("Inflectional:")
print("infect -> infected")
print("The suffix '-ed' indicates grammatical tense.")
print("Removing it can help retrieve related word forms.")

print("\nDerivational:")
print("infect -> infection")
print("infect -> infectious")
print("These forms create different words and grammatical categories.")
print("Blind stemming can remove important biomedical meaning.")


# -------------------------------
# 6. FINAL CONCLUSION
# -------------------------------

print("\nCONCLUSION")
print("----------")

print("Porter Stemmer is useful for simple word normalization.")
print("However, it does not understand the actual meaning of words.")
print("For biomedical search, blindly stemming all words can")
print("remove important derivational information.")
print("Lemmatization or domain-specific morphological analysis")
print("is therefore a better approach.")
```

```python
#QUESTION 2

import re

# Social media sample data
text = """
The happiest customer gave an unbelievable review.
Running customers liked the reordering option.
Smartphones are useful.
The unreadable message was deleted.
"""

# Words to test
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Words that contain prefix/suffix combinations
test_words = [
    "happiest",
    "unbelievable",
    "running",
    "reordering",
    "smartphones",
    "unreadable"
]


# Original simple parser
def old_parser(word):

    parts = []

    if word.startswith("un"):
        parts.append("un")
        word = word[2:]

    elif word.startswith("re"):
        parts.append("re")
        word = word[2:]

    if word.endswith("ing"):
        parts.append("ing")
        word = word[:-3]

    elif word.endswith("est"):
        parts.append("est")
        word = word[:-3]

    elif word.endswith("able"):
        parts.append("able")
        word = word[:-4]

    elif word.endswith("s"):
        parts.append("s")
        word = word[:-1]

    return word, parts


# Improved parser
def new_parser(word):

    original = word
    prefixes = []
    suffixes = []

    # Prefix transitions
    if word.startswith("un"):
        prefixes.append("un")
        word = word[2:]

    if word.startswith("re"):
        prefixes.append("re")
        word = word[2:]

    # Suffix transitions
    if word.endswith("iest"):
        suffixes.append("iest")
        word = word[:-4]

    elif word.endswith("ing"):
        suffixes.append("ing")
        word = word[:-3]

    elif word.endswith("able"):
        suffixes.append("able")
        word = word[:-4]

    elif word.endswith("s"):
        suffixes.append("s")
        word = word[:-1]

    return word, prefixes, suffixes


print("FINITE-STATE MORPHOLOGICAL PARSER")
print("----------------------------------")

print("\nOriginal Parser:")

for word in test_words:

    root, affixes = old_parser(word)

    print(
        word,
        "-> Root:",
        root,
        "Affixes:",
        affixes
    )


print("\nImproved Parser:")

for word in test_words:

    root, prefixes, suffixes = new_parser(word)

    print(
        word,
        "-> Root:",
        root,
        "Prefixes:",
        prefixes,
        "Suffixes:",
        suffixes
    )


# Accuracy calculation
expected = {
    "happiest": "happy",
    "unbelievable": "believe",
    "running": "runn",
    "reordering": "order",
    "smartphones": "smartphone",
    "unreadable": "read"
}

correct_old = 0
correct_new = 0

for word in test_words:

    old_root, old_affixes = old_parser(word)
    new_root, prefixes, suffixes = new_parser(word)

    if old_root == expected[word]:
        correct_old += 1

    if new_root == expected[word]:
        correct_new += 1


old_accuracy = correct_old / len(test_words) * 100
new_accuracy = correct_new / len(test_words) * 100

print("\nAccuracy:")
print("Before correction =", old_accuracy, "%")
print("After correction  =", new_accuracy, "%")


print("\nComplexity:")
print("The parser checks prefixes and suffixes sequentially.")
print("For a word of length n, the basic processing is approximately O(n).")
```

```python
#QUESTION 3

import time
import re

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import nltk

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Download WordNet
nltk.download("wordnet")
nltk.download("omw-1.4")


# Load two news categories
categories = [
    "comp.sys.ibm.pc.hardware",
    "rec.sport.baseball"
]

train = fetch_20newsgroups(
    subset="train",
    categories=categories,
    remove=("headers", "footers", "quotes")
)

test = fetch_20newsgroups(
    subset="test",
    categories=categories,
    remove=("headers", "footers", "quotes")
)


stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()


# No stemming
def normal_text(text):

    return text.lower()


# Porter stemming
def stem_text(text):

    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    return " ".join(
        stemmer.stem(word)
        for word in words
    )


# Lemmatization
def lemma_text(text):

    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    return " ".join(
        lemmatizer.lemmatize(word)
        for word in words
    )


def classify(train_text, test_text, name):

    start = time.time()

    vectorizer = TfidfVectorizer(
        max_features=10000
    )

    X_train = vectorizer.fit_transform(train_text)
    X_test = vectorizer.transform(test_text)

    model = LogisticRegression(
        max_iter=1000
    )

    model.fit(X_train, train.target)

    prediction = model.predict(X_test)

    accuracy = accuracy_score(
        test.target,
        prediction
    )

    vocabulary = len(
        vectorizer.get_feature_names_out()
    )

    elapsed = time.time() - start

    print("\n", name)
    print("----------------------")
    print("Vocabulary Size:", vocabulary)
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print("Processing Time:", round(elapsed, 2), "seconds")


print("NEWS CLASSIFICATION ERROR ANALYSIS")
print("----------------------------------")

# Original
train_original = [
    normal_text(x) for x in train.data
]

test_original = [
    normal_text(x) for x in test.data
]

classify(
    train_original,
    test_original,
    "WITHOUT STEMMING"
)


# Porter stemming
train_stem = [
    stem_text(x) for x in train.data
]

test_stem = [
    stem_text(x) for x in test.data
]

classify(
    train_stem,
    test_stem,
    "PORTER STEMMING"
)


# Lemmatization
train_lemma = [
    lemma_text(x) for x in train.data
]

test_lemma = [
    lemma_text(x) for x in test.data
]

classify(
    train_lemma,
    test_lemma,
    "LEMMATIZATION"
)


# Individual word analysis
print("\nPORTER STEMMING ERROR ANALYSIS")
print("------------------------------")

words = [
    "organization",
    "organizer",
    "organizing",
    "organized",
    "organization's"
]

for word in words:

    clean = re.sub(
        r'[^a-zA-Z]',
        '',
        word
    )

    print(
        word,
        "->",
        stemmer.stem(clean)
    )
```

```python
#QUESTION 6

# Simple morphological plural parser

irregular_plurals = {
    "children": "child",
    "men": "man",
    "women": "woman",
    "mice": "mouse",
    "feet": "foot",
    "teeth": "tooth",
    "people": "person"
}


def parser(word):

    # Irregular plurals
    if word in irregular_plurals:

        return (
            irregular_plurals[word],
            "Plural Noun"
        )


    # Words ending in ies
    if word.endswith("ies"):

        return (
            word[:-3] + "y",
            "Plural Noun"
        )


    # Words ending in es
    if word.endswith("es"):

        return (
            word[:-2],
            "Plural Noun"
        )


    # Regular plural ending in s
    if word.endswith("s"):

        return (
            word[:-1],
            "Plural Noun"
        )


    # Singular
    return (
        word,
        "Singular Noun"
    )


words = [
    "cars",
    "boxes",
    "cities",
    "children",
    "books",
    "dogs",
    "men",
    "women",
    "mice",
    "cat"
]


print("MORPHOLOGICAL PLURAL PARSER")
print("----------------------------")

for word in words:

    root, tag = parser(word)

    print(
        f"{word:10} -> {root:10} -> {tag}"
    )
```

```python
#QUESTION 7

from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression

stemmer = PorterStemmer()


documents = [
    "running runners runs",
    "studies studied studying",
    "organization organized organizer",
    "running studies organization",
    "runners organized studies"
]

labels = [
    0,
    1,
    2,
    0,
    1
]


# -------------------------------
# BEFORE CORRECTION
# -------------------------------

vectorizer1 = CountVectorizer()

X1 = vectorizer1.fit_transform(
    documents
)

vocab_before = (
    vectorizer1.get_feature_names_out()
)

print("BEFORE CORRECTION")
print("------------------")

print("Vocabulary:")
print(vocab_before)

print(
    "Vocabulary Size:",
    len(vocab_before)
)


# -------------------------------
# AFTER CORRECTION
# -------------------------------

def stem_document(text):

    words = text.lower().split()

    return " ".join(
        stemmer.stem(word)
        for word in words
    )


stemmed_documents = []

for document in documents:

    stemmed_documents.append(
        stem_document(document)
    )


print("\nSTEMMED DOCUMENTS")
print("------------------")

for document in stemmed_documents:

    print(document)


# Vectorization AFTER stemming
vectorizer2 = CountVectorizer()

X2 = vectorizer2.fit_transform(
    stemmed_documents
)

vocab_after = (
    vectorizer2.get_feature_names_out()
)

print("\nAFTER CORRECTION")
print("----------------")

print("Vocabulary:")
print(vocab_after)

print(
    "Vocabulary Size:",
    len(vocab_after)
)


# -------------------------------
# CLASSIFICATION
# -------------------------------

model = LogisticRegression(
    max_iter=1000
)

model.fit(X2, labels)

accuracy = model.score(
    X2,
    labels
)

print("\nCLASSIFICATION")
print("--------------")

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)


# -------------------------------
# COMPARISON
# -------------------------------

print("\nCOMPARISON")
print("----------")

print(
    "Vocabulary Before:",
    len(vocab_before)
)

print(
    "Vocabulary After:",
    len(vocab_after)
)

print(
    "Vocabulary Reduction:",
    len(vocab_before) -
    len(vocab_after)
)
```
