```python
# Question 1
# Rule-Based Morphological Processing

words = ["analyzing", "analysis", "analytical"]

def analyze_word(word):
    if word.endswith("ing"):
        root = word[:-3]
        if root.endswith("z"):
            root = root[:-1] + "ze"
        suffix = "-ing"
        form = "Inflectional"
        normalized = "analyze"

    elif word.endswith("sis"):
        root = "analyze"
        suffix = "-sis"
        form = "Derivational"
        normalized = "analyze"

    elif word.endswith("ical"):
        root = "analyze"
        suffix = "-ical"
        form = "Derivational"
        normalized = "analyze"

    else:
        root = word
        suffix = "-"
        form = "Base"
        normalized = word

    return [word, root, suffix, form, normalized]

print("="*80)
print("{:<15}{:<15}{:<15}{:<18}{:<15}".format(
    "Original","Root","Affix","Transformation","Normalized"))
print("="*80)

for w in words:
    result = analyze_word(w)
    print("{:<15}{:<15}{:<15}{:<18}{:<15}".format(*result))
```

```python
# Question 2
# Rule-Based Morphological Parser

words = ["disagree", "agreement", "agreeable"]

def parse(word):

    prefix = "-"
    suffix = "-"
    semantic = ""

    if word.startswith("dis"):
        prefix = "dis-"
        root = "agree"
        category = "Derivational"
        semantic = "Negation of agreement"

    elif word.endswith("ment"):
        root = "agree"
        suffix = "-ment"
        category = "Derivational"
        semantic = "State or result of agreeing"

    elif word.endswith("able"):
        root = "agree"
        suffix = "-able"
        category = "Derivational"
        semantic = "Capable of being agreed with"

    else:
        root = word
        category = "Base"
        semantic = "Original word"

    normalized = "agree"

    return [word, prefix, root, suffix,
            category, semantic, normalized]

print("="*120)
print("{:<15}{:<10}{:<12}{:<10}{:<18}{:<35}{:<15}".format(
    "Original","Prefix","Root","Suffix",
    "Category","Semantic Meaning","Normalized"))
print("="*120)

for w in words:
    print("{:<15}{:<10}{:<12}{:<10}{:<18}{:<35}{:<15}".format(*parse(w)))
```

```python
# Question 3
# Morphology-Based Normalization

words = ["govern", "government", "governance"]

def normalize(word):

    if word == "govern":
        root = "govern"
        affix = "-"
        level = "Base"

    elif word.endswith("ment"):
        root = "govern"
        affix = "-ment"
        level = "Level-1 Derivation"

    elif word.endswith("ance"):
        root = "govern"
        affix = "-ance"
        level = "Level-1 Derivation"

    normalized = "govern"

    return [word, root, affix, level,
            normalized, normalized]

print("="*90)
print("{:<15}{:<15}{:<15}{:<25}{:<15}{:<15}".format(
    "Original","Root","Affix",
    "Derivation Level","Normalized","Final"))
print("="*90)

for w in words:
    print("{:<15}{:<15}{:<15}{:<25}{:<15}{:<15}".format(*normalize(w)))
```

```python
# Question 4
# Morphological Parsing and Normalization

words = ["activate", "activation", "reactivation"]

def parser(word):

    prefix = "-"
    suffix = "-"
    sequence = ""
    meaning = ""

    if word == "activate":
        root = "act"
        suffix = "-ivate"
        sequence = "act → activate"
        meaning = "Verb"

    elif word == "activation":
        root = "act"
        suffix = "-ivation"
        sequence = "act → activate → activation"
        meaning = "Verb → Noun"

    elif word == "reactivation":
        prefix = "re-"
        root = "act"
        suffix = "-ivation"
        sequence = "act → activate → activation → reactivation"
        meaning = "Again activate"

    normalized = "act"

    return [word, prefix, root, suffix,
            sequence, normalized, meaning]

print("="*140)
print("{:<15}{:<10}{:<10}{:<15}{:<45}{:<15}{:<20}".format(
    "Original","Prefix","Root","Suffix",
    "Derivational Sequence","Normalized","Meaning"))
print("="*140)

for w in words:
    print("{:<15}{:<10}{:<10}{:<15}{:<45}{:<15}{:<20}".format(*parser(w)))
```

```python
# Question 5
# Inflectional Morphology-Based Normalization

words = ["create", "creates", "creating"]

def process(word):

    suffix = "-"
    grammar = ""

    if word == "create":
        root = "create"
        grammar = "Base Form"

    elif word.endswith("s"):
        root = "create"
        suffix = "-s"
        grammar = "Third-Person Singular"

    elif word.endswith("ing"):
        root = "create"
        suffix = "-ing"
        grammar = "Present Participle"

    normalized = "create"

    return [word, suffix, grammar,
            root, normalized, normalized]

print("="*100)
print("{:<15}{:<10}{:<25}{:<15}{:<15}{:<15}".format(
    "Original","Suffix","Grammar",
    "Root","Normalized","Final"))
print("="*100)

for w in words:
    print("{:<15}{:<10}{:<25}{:<15}{:<15}{:<15}".format(*process(w)))
```
