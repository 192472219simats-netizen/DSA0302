```python
# CASE STUDY 2 - AI Powered Customer Support Chatbot

# Input sentences
s1 = "Book a flight ticket now."
s2 = "This book is interesting."

# ---------------------------------------------------
# 1. Penn Treebank POS Tagging (manual assignment)
# ---------------------------------------------------

tags_s1 = {
    "Book": "VB",
    "a": "DT",
    "flight": "NN",
    "ticket": "NN",
    "now": "RB"
}

tags_s2 = {
    "This": "DT",
    "book": "NN",
    "is": "VBZ",
    "interesting": "JJ"
}

print("Sentence 1 POS Tags:")
for w, t in tags_s1.items():
    print(f"{w:12} -> {t}")

print("\nSentence 2 POS Tags:")
for w, t in tags_s2.items():
    print(f"{w:12} -> {t}")

# ---------------------------------------------------
# 2. HMM Probability for 'book' as Verb
# ---------------------------------------------------

P_book_VB = 0.6
P_book_NN = 0.4
P_start_VB = 0.5
P_start_NN = 0.5

# Likelihoods
likelihood_VB = P_start_VB * P_book_VB
likelihood_NN = P_start_NN * P_book_NN

print("\nHMM Likelihoods:")
print("P(Start -> VB) * P(book | VB) =", likelihood_VB)
print("P(Start -> NN) * P(book | NN) =", likelihood_NN)

if likelihood_VB > likelihood_NN:
    print("Predicted Tag for 'Book': VB (Verb)")
else:
    print("Predicted Tag for 'Book': NN (Noun)")

# ---------------------------------------------------
# 3. Rule-based vs HMM comparison
# ---------------------------------------------------

print("\nRule-based tagging:")
print("- Uses grammar rules and context patterns")
print("- Example: sentence starts with an action word => VB")

print("\nStochastic (HMM) tagging:")
print("- Uses probabilities from training data")
print("- Chooses the tag with highest likelihood")

# ---------------------------------------------------
# 4. Chatbot performance notes
# ---------------------------------------------------

print("\nRole of standardized POS tags:")
print("- Detects user intent (booking, cancellation, inquiry)")
print("- Improves response generation")
print("- Reduces ambiguity")
print("- Enhances overall chatbot accuracy")
```

```python
#CASE STUDY 3

import math

# ---------------------------------------------------
# Input sentence and initial tags
# ---------------------------------------------------

words = ["economic", "growth", "increases", "employment"]
tags = ["JJ", "NN", "NNS", "NN"]

print("Initial POS Tags:")
for w, t in zip(words, tags):
    print(f"{w:12}/{t}")

# ---------------------------------------------------
# 1. Apply Brill transformation rule
# Rule: Change NNS -> VBZ if previous tag is NN
# ---------------------------------------------------

corrected_tags = tags.copy()

for i in range(1, len(tags)):
    if tags[i] == "NNS" and tags[i-1] == "NN":
        corrected_tags[i] = "VBZ"

print("\nCorrected POS Tags:")
for w, t in zip(words, corrected_tags):
    print(f"{w:12}/{t}")

# ---------------------------------------------------
# 2. Frequency Distribution
# ---------------------------------------------------

freq = {
    "economic": 120,
    "growth": 450,
    "increases": 210,
    "employment": 380
}

total = sum(freq.values())

print("\nWord Frequency Distribution:")
for word, count in freq.items():
    prob = count / total
    print(f"{word:12}: {count:3}  Probability = {prob:.3f}")

# ---------------------------------------------------
# 3. Entropy Before and After Correction
# Assume ambiguity for 'increases'
# Before: NNS=0.55, VBZ=0.45
# After : VBZ=0.95, NNS=0.05
# ---------------------------------------------------

def entropy(probs):
    return -sum(p * math.log2(p) for p in probs if p > 0)

H_before = entropy([0.55, 0.45])
H_after = entropy([0.95, 0.05])

print("\nEntropy Analysis:")
print("Before correction =", round(H_before, 4), "bits")
print("After correction  =", round(H_after, 4), "bits")

# ---------------------------------------------------
# 4. Accuracy Improvement Explanation
# ---------------------------------------------------

print("\nTransformation-Based Tagging improves accuracy by:")
print("- Detecting context-sensitive errors")
print("- Applying learned correction rules")
print("- Reducing ambiguity")
print("- Lowering entropy and increasing confidence")
```
